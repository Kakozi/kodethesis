```python
import pandas as pd

input_file = "AnonDB_zub.csv"
output_file = "AnonDB_zub_60.csv"

df = pd.read_csv(input_file, delimiter=";")

if "Num_Cells" not in df.columns:
    raise ValueError("Kolom 'Num_Cells' tidak ditemukan di file CSV.")

# Ubah Num_Cells menjadi angka
df["Num_Cells"] = pd.to_numeric(df["Num_Cells"], errors="coerce")

# Filter 60 cells
df_60 = df[df["Num_Cells"] == 60].copy()

# Simpan hasil
df_60.to_csv(output_file, index=False)

print(f"File berhasil dibuat: {output_file}")
print(f"Jumlah data awal: {len(df)}")
print(f"Jumlah data dengan Num_Cells = 60: {len(df_60)}")
print(f"Jumlah data dibuang: {len(df) - len(df_60)}")
```

```python
import pandas as pd

input_file = "AnonDB_sdm_sandia.csv"
output_file = "AnonDB_sandia_60.csv"

# Baca semua kolom sebagai string dulu supaya kode indeks tidak berubah jadi float
df = pd.read_csv(input_file, delimiter=";", dtype=str)

# Hapus kolom Unnamed kalau memang hanya kolom index kosong dari CSV sebelumnya
unnamed_cols = [col for col in df.columns if col.startswith("Unnamed")]

print("Kolom Unnamed yang ditemukan:")
print(unnamed_cols)

# Kalau kolom Unnamed itu tidak diperlukan, hapus
df = df.drop(columns=unnamed_cols)

# Pastikan kolom Num_Cells ada
if "Num_Cells" not in df.columns:
    raise ValueError("Kolom 'Num_Cells' tidak ditemukan di file CSV.")

# Bersihkan Num_Cells lalu ubah ke numeric hanya untuk filtering
num_cells_numeric = pd.to_numeric(df["Num_Cells"], errors="coerce")

# Filter hanya Num_Cells = 60
df_60 = df[num_cells_numeric == 60].copy()

# Simpan hasil
df_60.to_csv(output_file, index=False)

print(f"File berhasil dibuat: {output_file}")
print(f"Jumlah data awal: {len(df)}")
print(f"Jumlah data dengan Num_Cells = 60: {len(df_60)}")
print(f"Jumlah data dibuang: {len(df) - len(df_60)}")
```

```python
import pandas as pd
from pathlib import Path

input_file = "AnonDB_Zub_96.csv"
output_file = "AnonDB_Zub_96.csv"

df = pd.read_csv(input_file, delimiter=",")

df["module_code"] = (
    df["ELHPath"]
    .astype(str)
    .apply(lambda x: Path(x).stem)          # ambil nama file tanpa .tiff
    .str.replace(r"_(20|80)$", "", regex=True)  # hapus suffix _80 atau _20
)

df.to_csv(output_file, index=False)

print(f"Saved to {output_file}")
print(df[["ELHPath", "module_code"]].head())

#df.to_csv("AnonDB_Zub_72_96.csv", index=False)
```

```python
import pandas as pd
from sklearn.model_selection import GroupShuffleSplit


# ============================================================
# SETTINGS
# ============================================================
INPUT_CSV = "AnonDB_80_10_10_seed42.csv"
OUTPUT_CSV = "AnonDB_test_make34_38_trainval_seed42.csv"

TEST_MAKES = [34, 38]
VALIDATION_FRACTION = 0.10   # 10% dari data non-test
RANDOM_SEED = 42


# ============================================================
# READ DATA
# ============================================================
df = pd.read_csv(INPUT_CSV)

# Ensure numeric make
df["Make"] = pd.to_numeric(df["Make"], errors="coerce")

# Reset existing split
df["split"] = ""


# ============================================================
# FIXED TEST SET: MAKE 34 AND 38
# ============================================================
test_mask = df["Make"].isin(TEST_MAKES)

df.loc[test_mask, "split"] = "test"


# ============================================================
# TRAIN / VALIDATION FROM ALL REMAINING MAKES
# Split by Mod_ID to avoid module leakage
# ============================================================
remaining_df = df.loc[~test_mask].copy()

splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=VALIDATION_FRACTION,
    random_state=RANDOM_SEED,
)

train_idx, val_idx = next(
    splitter.split(
        remaining_df,
        groups=remaining_df["Mod_ID"],
    )
)

train_original_indices = remaining_df.iloc[train_idx].index
val_original_indices = remaining_df.iloc[val_idx].index

df.loc[train_original_indices, "split"] = "train"
df.loc[val_original_indices, "split"] = "val"


# ============================================================
# VALIDATION CHECKS
# ============================================================
print("\nRow counts:")
print(df["split"].value_counts())

print("\nRow percentages:")
print(
    (df["split"].value_counts(normalize=True) * 100)
    .round(2)
    .astype(str) + "%"
)

print("\nMakes in each split:")
for split_name in ["train", "val", "test"]:
    makes = sorted(
        df.loc[df["split"] == split_name, "Make"]
        .dropna()
        .unique()
        .tolist()
    )

    print(f"{split_name}: {makes}")


# Check fixed test makes
assert set(
    df.loc[df["split"] == "test", "Make"].dropna().unique()
) == set(TEST_MAKES)

assert not df.loc[
    df["Make"].isin(TEST_MAKES),
    "split"
].ne("test").any()


# Check Mod_ID leakage
train_modules = set(df.loc[df["split"] == "train", "Mod_ID"])
val_modules = set(df.loc[df["split"] == "val", "Mod_ID"])
test_modules = set(df.loc[df["split"] == "test", "Mod_ID"])

assert train_modules.isdisjoint(val_modules)
assert train_modules.isdisjoint(test_modules)
assert val_modules.isdisjoint(test_modules)

assert not df["split"].eq("").any()

print("\nNo Mod_ID leakage between train, validation, and test.")


# ============================================================
# SAVE
# ============================================================
df.to_csv(
    OUTPUT_CSV,
    index=False,
)

print("\nSaved to:", OUTPUT_CSV)
```
