```python
import os
import numpy as np
from PIL import Image


# ============================================================
# INPUT AND OUTPUT
# ============================================================
INPUT_TIFF = "36khususkitapotong/2533_36_1_08062019_80_wrp.tiff"
OUTPUT_DIR = "2533_36_1_cells"

os.makedirs(OUTPUT_DIR, exist_ok=True)


# ============================================================
# READ IMAGE
# ============================================================
image = Image.open(INPUT_TIFF)
image_array = np.array(image)

height, width = image_array.shape

print("Image shape:", image_array.shape)
print("Data type:", image_array.dtype)


# ============================================================
# MANUALLY DETERMINED CELL BOUNDARIES
# ============================================================
# Outer boundaries plus internal separators
x_boundaries = [
    100,
    531,
    964,
    1400,
    1837,
    2278,
    2718,
    3160,
    3601,
    4042,
    4480,
    4917,
    5370
]

y_boundaries = [
    128,
    569,
    1010,
    1447,
    1891,
    2334,
    2777
]


# ============================================================
# OPTIONAL MARGIN
# ============================================================
# Positive margin removes some black gap around each cell.
# Try values between 5 and 20 pixels.
margin_x = 8
margin_y = 8


# ============================================================
# CROP ALL CELLS
# ============================================================
cell_number = 1

for row in range(6):
    for col in range(12):

        x1 = x_boundaries[col] + margin_x
        x2 = x_boundaries[col + 1] - margin_x

        y1 = y_boundaries[row] + margin_y
        y2 = y_boundaries[row + 1] - margin_y

        cell = image_array[y1:y2, x1:x2]

        output_name = (
            f"2533_36_1_08062019"
            f"_cell_{cell_number:02d}"
            f"_r{row + 1:02d}"
            f"_c{col + 1:02d}.tiff"
        )

        output_path = os.path.join(OUTPUT_DIR, output_name)

        Image.fromarray(cell).save(output_path)

        print(
            f"Saved cell {cell_number:02d}: "
            f"row={row + 1}, col={col + 1}, "
            f"shape={cell.shape}"
        )

        cell_number += 1


print(f"\nFinished. Saved {cell_number - 1} cells to:")
print(OUTPUT_DIR)
```

```python
import matplotlib.pyplot as plt


fig, ax = plt.subplots(figsize=(18, 10))

ax.imshow(image_array, cmap="gray")

for x in x_boundaries:
    ax.axvline(x, linewidth=1)

for y in y_boundaries:
    ax.axhline(y, linewidth=1)

ax.set_xlim(0, image_array.shape[1])
ax.set_ylim(image_array.shape[0], 0)
ax.set_title("Adjusted 6 × 12 Cell Cropping Grid")

plt.tight_layout()
plt.show()
```
